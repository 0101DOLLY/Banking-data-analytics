Load final_df dataset on mysql

 Now Import the 'mysql.connector' library

```python
import mysql.connector 
```

Connect Python with MySQL Server

```python
conn = mysql.connector.connect(
    host="localhost",
    port=3306,
    user="root",
    password="1234",
    database="bank_df"
)

if conn.is_connected():
    print("Connected to bank_df successfully!")
```

Create a Cursor and Check Available Tables


```python
cursor = conn.cursor()

cursor.execute("SHOW TABLES")

for table in cursor:
    print(table)
```

Upload final_df to MySQL

```python
from sqlalchemy import create_engine

engine = create_engine(
    "mysql+mysqlconnector://root:1234@localhost:3306/bank_df"
)

final_df.to_sql(
    "final_df",
    con=engine,
    if_exists="replace",
    index=False
)

print("final_df uploaded successfully!")
```
